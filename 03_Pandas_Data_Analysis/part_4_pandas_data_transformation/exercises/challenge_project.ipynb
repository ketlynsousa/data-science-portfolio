## STAGE 3

### - Challenge Project 1: Online Sales

In stage 2, we worked on transforming numerical data. Now, we can address textual values.

Due to an instability on the company's website, issues arose with customer names during the saving process. This resulted in a `"Cliente"` (costumer) column containing a mix of uppercase and lowercase letters, numbers, and other characters.

Given this, manipulate the text in the `"Cliente"` (costumer) column so that the final output consists of customer names in all lowercase letters, with all special characters and numbers removed.

```python
sales_data['Cliente']
```

```python
sales_data['Cliente'] = sales_data['Cliente'].str.lower()
```

```python
sales_data.head(3)
```

```python
sales_data['Cliente'] = sales_data['Cliente'].str.replace('[^a-z ]', '', regex=True)
sales_data['Cliente']
```

```python
sales_data.head()
```

### - Challenge Project 2: Condominium Management

As in the first challenge project, we were working with all columns except those involving dates.

In the `payment_dates` (*datas_de_pagamento*) and `agreed_payment_dates` (*datas_combinadas_pagamento*) columns, the dates are in 'day/month/year' (dd/mm/yyyy) format. Convert this data to the `datetime` type and find a way to visualize a subset of the data that contributes to the objectives of the context in which this information is situated.

If you do not recall the problem from the second challenge project, I have included the description of the scenario below to help you find the necessary information:

Managing a condominium complex is a task that requires a great deal of attention and organization. Among the various management responsibilities is the collection of rent from tenants. To ensure the financial health of the property, it is essential that these payments are made regularly and on time. However, we know that this does not always happen. With this in mind, I am proposing a data-handling challenge aimed at analyzing rent payment delays among some residents of a hypothetical condominium.

```python
condominium_data.head()
```

```python
condominium_data.info()
```

```python
condominium_data['datas_combinadas_pagamento'] = pd.to_datetime(condominium_data['datas_combinadas_pagamento'], format='%d/%m/%Y')
condominium_data['datas_de_pagamento'] = pd.to_datetime(condominium_data['datas_de_pagamento'], format='%d/%m/%Y')
```

```python
condominium_data.info()
```

```python
condominium_data.head()
```

```python
condominium_data['payment_delay'] = (condominium_data['datas_de_pagamento'] - condominium_data['datas_combinadas_pagamento']).dt.days
```

```python
condominium_data.head()
```

```python
subset = condominium_data.groupby(['apartamento'])['payment_delay'].mean()
subset
```
