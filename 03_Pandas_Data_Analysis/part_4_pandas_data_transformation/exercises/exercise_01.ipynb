During the lessons, we performed tokenization on two columns from our hospitality dataset: `descricao_local` (location description) and `comodidades` (amenities). However, the `descricao_vizinhanca` (neighborhood description) column remains, and it is also worth processing.

Therefore, for this activity, I propose that you carry out the tokenization process for the `descricao_vizinhanca` column found in the `dados_hospedagem.json` dataset.

```python
import pandas as pd
```

```python
data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_4_pandas_data_transformation/resources/normalized_data_hospitality_v3.json')
data['descricao_vizinhanca']
```

```python
data['descricao_vizinhanca'] = data['descricao_vizinhanca'].str.lower()
```

```python
data['descricao_vizinhanca'] = data['descricao_vizinhanca'].replace('[^a-zA-Z0-9\-\']', ' ', regex=True)
data['descricao_vizinhanca']
```

```python
data['descricao_vizinhanca'] = data['descricao_vizinhanca'].str.replace('(?<!\w)-(?!\w)', ' ', regex=True)
data['descricao_vizinhanca']
```

```python
data['descricao_vizinhanca'] = data['descricao_vizinhanca'].str.split()
data['descricao_vizinhanca']
```

```python
data.to_json('normalized_data_hospitality_v4.json', index=False)
```
