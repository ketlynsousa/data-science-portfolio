In this lesson, we learned how to manipulate temporal data using the `datetime` type. We explored how to convert a column to `datetime` and subsequently work with that data. However, not all the data in the `moveis_disponiveis.json` dataset has been processed yet.

Throughout this course, we discovered how to transform and work with numerical values—for instance, by extracting numbers embedded within text and converting them into a numerical data type, such as `int64` or `float64`.

With this in mind, for this activity, I would like you to convert the data in the `preco` (price) column of the `moveis_disponiveis.json` dataset into the `float64` numerical type. Keep in mind that, before doing so, you must fill in any missing (null) values ​​in that column. A suggestion: replace them with the string `'0.0'`.

```python
import pandas as pd
import numpy as np
```

```python
dt_data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_4_pandas_data_transformation/resources/moveis_disponiveis_v2.json')
```

```python
dt_data.head()
```

```python
dt_data.info()
```

```python
dt_data['preco'].fillna('0.0', inplace=True)
```

```python
dt_data['preco'] = dt_data['preco'].apply(lambda x: x.replace('$', '').replace(',', '.'))
dt_data['preco']
```

```python
dt_data['preco'] = dt_data['preco'].apply(lambda x: x.replace('.', '', x.count('.') - 1) if isinstance(x, str) else x)
dt_data['preco'] = dt_data['preco'].astype(np.float64)
dt_data['preco']
```

```python
dt_data.info()
```

```python
dt_data.to_json('moveis_disponiveis_v3.json', index=False)
```
