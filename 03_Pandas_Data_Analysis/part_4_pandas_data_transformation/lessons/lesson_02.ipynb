# Ungrouping Data

Previously, we examined our data and observed that some columns contain data grouped into lists. To make this data easier to access so that we can work with it, we need to ungroup it.

The columns containing the data grouped into lists are:
- descrição_local (location_description)
- descricao_vizinhanca (neighborhood_description)
- quantidade_banheiros (number_of_bathrooms)
- quantidade_quartos (number_of_bedrooms)
- quantidade_camas (number_of_beds)
- modelo_cama (bed_model)
- comodidades (amenities)
- taxa_deposito (deposit_fee)
- taxa_limpeza (cleaning_fee)
- preco (price)

```python
import pandas as pd
```

___
Normalized json set of data
___

```python
data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_4_pandas_data_transformation/resources/normalized_data_hospitality.json')
data.head()
```

### Collecting column names

To ungroup this data, we can use the `.explode()` method, which separates the information into rows. It requires us to provide the names of the columns to be transformed, so the first step is to collect those names.

```python
columns_list = list(data.columns)
columns_list
```

### Usando o `.explode()` para desagrupar dados

The columns "avaliacao_geral", "experiencia_local", and "max_hospedes" are not grouped into lists. Therefore, we will use `.explode()` starting from "descricao_local", which corresponds to index 3.

```python
data = data.explode(columns_list[3:])
data
```

### Resetting the indices to sort the data

Note, however, that the index of the new dataframe is incorrect, as zero—along with other numbers—is repeated multiple times. We therefore need to reset the indices.

To do this, we will use the `.reset_index()` method, passing the parameters `inplace=True` and `drop=True`. The first parameter replaces the indices in the dataframe, while the second relates to the creation of a column containing the old indices; by setting it to `True`, we prevent this column from being created.

```python
data.reset_index(inplace=True, drop=True)
data.head()
```

# Checking data types

Now that we have ungrouped the data in the columns, we can check what type of data we are working with in each column.

```python
data.info()
```

### Handling numerical data: converting to integer type

The `.astype()` method allows us to convert columns to a specific data type. To do this, we reference our dataframe—let’s call it `dados`—and specify the column we want to convert within square brackets, for example: `dados['max_hospedes']`.

```python
import numpy as np
```

```python
data['max_hospedes'] = data['max_hospedes'].astype(np.int64)
```

```python
data.info()
```

```python
numerics_columns = ['quantidade_banheiros', 'quantidade_quartos', 'quantidade_camas']
```

```python
data[numerics_columns] = data[numerics_columns].astype('int64')
```

```python
data.info()
```

But will `.astype()` work on data that contains a dollar sign (`$`) and a comma (`,`)? Let’s find out!

```python
data['preco'] = data['preco'].astype(np.float64)
```

We get an error because it is not possible to apply the conversion to data that contains the dollar sign character.

```python
data['preco']
```

### Separating one column in a dataframe
Using the `.apply()` method to manipulate data in a single column or row.

```python
data['preco'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
data['preco'] = data['preco'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
data['preco'] = data['preco'].astype(np.float64)
```

```python
data.info()
```

### Separate two columns or more in a dataframe
Using the `.applymap()` method to manipulate multiple data in a column or row.

```python
data[['taxa_deposito', 'taxa_limpeza']]
```

```python
data[['taxa_deposito', 'taxa_limpeza']].applymap(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
data[['taxa_deposito', 'taxa_limpeza']] = data[['taxa_deposito', 'taxa_limpeza']].applymap(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
data[['taxa_deposito', 'taxa_limpeza']] = data[['taxa_deposito', 'taxa_limpeza']].astype(np.float64)
```

```python
data.info()
```

```python
data.to_json('normalized_data_hospitality_v2.json', index=False)
```
