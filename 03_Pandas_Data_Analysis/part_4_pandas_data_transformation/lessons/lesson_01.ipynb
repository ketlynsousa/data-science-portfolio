```python
import pandas as pd
```

```python
data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_4_pandas_data_transformation/resources/dados_hospedagem.json')
data.head()
```

We can see what our `dados_hospedagem.json` file looks like, and it’s clear that the data is nested in a dictionary format. This makes it difficult to interpret. Therefore, we will need to unnest the data so that we can better examine the values ​​and results contained in this dataframe.

```python
data = pd.json_normalize(data['info_moveis'])
data
```

Finally, we can save a normalized version of the data, as we will be using this dataframe multiple times to solve our problem.

```python
data.to_json('normalized_data_hospitality.json', index=False)
```
