### Regex: removing special characters

To replace these special characters, we will use `str` along with another string method, `replace()`.

Let's understand its structure.:


- `^`: excludes the following values ​​(they will not be selected)
- `a-z`: all letters from "a" to "z", lowercase;
- `A-Z`: all letters from "A" to "Z", uppercase;
- `0-9`: all numbers from 0 to 9;
- `\-`: hyphen;
- `\'`: apostrophe.

This entire expression means: "select all characters, except for letters (uppercase or lowercase), numbers, hyphens, and apostrophes."

```python
data['descricao_local'] = data['descricao_local'].str.replace('[^a-zA-Z0-9\-\']', ' ', regex=True)
data['descricao_local']
```

### Removing hyphens outside of compound words

- (`?<!\w`): a negative lookbehind that checks whether there is a word character before the hyphen or just a whitespace character, resulting in "true" if there is nothing there;

- (`?!\w`): a negative lookahead that checks whether there is a word character after the hyphen.

```python
data['descricao_local'] = data['descricao_local'].str.replace('(?<!\w)-(?!\w)', ' ', regex=True)
data['descricao_local']
```

# String tokenization

To create tokens, simply separate each of the words as elements of a list.

```python
data['descricao_local'] = data['descricao_local'].str.split()
data['descricao_local']
```

```python
data.head()
```

### Data processing treatment

We have another column for `comodidades` which resembles tokenization but isn't well-structured as a list.

So, we need to process it so that our tokenization is standardized throughout the entire dataframe.

- `{|` open brace;
- `}|` close brace;
- `"` quotation marks.
  - the vertical bars (|) represent a choice between two different patterns.

```python
data['comodidades'] = data['comodidades'].str.replace('\{|}|\"', '', regex=True)
data['comodidades']
```

In this case, we can keep the commas to use them as a separator for splitting the words into a list during our tokenization.

```python
data['comodidades'] = data['comodidades'].str.split(',')
data['comodidades']
```
