Import the file [**alunos.csv**](https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/desafios/alunos.csv) again to complete the challenges below.

```python
# Importando novamente os dados
import pandas as pd

url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/desafios/alunos.csv'

data = pd.read_csv(url)
data
```

### Challenge 1
Students participated in an extracurricular activity and earned extra points. These extra points correspond to 40% of each student's current grade. Based on this, create a column named "Pontos_extras" (Extra_points) that contains the extra points for each student—that is, 40% of their current grade.

```python
percentage = 40 / 100

data['Pontos_extras'] = data['Notas'].apply(lambda x: x * percentage)
data
```

### Challenge 2
Create an additional column named "Final_grades" that contains each student's grades with the extra points added.

```python
data['Final_grades'] = data['Notas'] + data['Pontos_extras']
data
```

### Challenge 3
Since the students received extra points, some who had not previously passed may now have passed. Based on this, create a column named "Aprovado_final" with the following values:

* True: if the student has passed (final grade must be greater than or equal to 6);
* False: if the student has failed (final grade must be less than 6).


```python
data['Aprovado_final'] = data['Final_grades'].apply(lambda x: True if x >= 6 else False)
data
```

### Challenge 4
Identify which students had not previously passed but did pass after the addition of extra points.

```python
data.query('Aprovado == False & Aprovado_final == True')
```

```python
selection = (data['Aprovado'] == False) & (data['Aprovado_final'] == True)
data[selection]
```
