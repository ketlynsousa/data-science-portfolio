## Challenge

Import the file [**alunos.csv**](https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/desafios/alunos.csv) again to complete the challenges below.

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/desafios/alunos.csv'
```

```python
data = pd.read_csv(url)
data.head()
```

### Challenge 1
Check whether the dataset contains null values ​​and, if so, handle these null values.

```python
data.isnull().sum()
```

```python
data = data.fillna(0)
```

```python
data.isnull().sum()
```

### Challenge 2
The students "Alice" and "Carlos" are no longer part of the class. Therefore, remove them from the database.

```python
students_to_remove = data.query('Nome == "Alice" or Nome == "Carlos"').index
students_to_remove
```

```python
data.drop(students_to_remove, axis=0, inplace=True)
```

```python
data
```

### Challenge 3
Apply a filter that selects only the students who passed

**One way to do it**

```python
data = data.query('Aprovado == True')
```

**Second way of doing it**

```python
selection = data['Aprovado'] == True
data[selection]
```

### Challenge 4
Save the DataFrame containing only the students who passed to a CSV file named "alunos_aprovados.csv"

```python
approved_students = data[selection]
```

```python
approved_students.to_csv('students_approved.csv', index=False)
```

```python
pd.read_csv('students_approved.csv')
```

### **Extra**
Upon reviewing the grades of the students who passed, we noticed that some scores were incorrect. The students who received a 7.0 actually had an extra point that was not accounted for. Therefore, replace the 7.0 grades in the dataset with 8.0. **Tip**: look into the **replace** method.

```python
approved_students.replace(7.0, 8.0)
```

```python
approved_students = approved_students.replace(7.0, 8.0)
```

```python
approved_students
```

```python
approved_students.to_csv('students_approved.csv', index=False)
```
