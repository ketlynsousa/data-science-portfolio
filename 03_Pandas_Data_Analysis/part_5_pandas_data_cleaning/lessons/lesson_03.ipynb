# Identification of duplicate data

Duplicate samples are rows that contain exactly the same values ​​in all columns. In other words, they are exact copies of other rows present in the dataset.

```python
import pandas as pd
```

```python
no_empty_data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/resources/churn_subset_without_empty_data_v1.json')
```

***
The Pandas library provides the `duplicated()` method, which returns a Boolean Series where `False` values ​​indicate that there is no duplication of this set of data, and `True` values ​​indicate that there is duplicate data.
***

### Identifying and handling duplicate data

```python
no_empty_data.duplicated()
```

```python
no_empty_data.duplicated().sum()
```

```python
duplicated_filter = no_empty_data.duplicated()
duplicated_filter
```

```python
no_empty_data[duplicated_filter]
```

With this analysis, we now know how many duplicate samples there are, which ones they are, and how to locate them within the dataset. But what should be done with these samples? They should be removed from the dataset.

There are three main reasons for removing duplicate samples:

- Model bias: If a dataset contains duplicate samples, a machine learning model might assign disproportionate importance to these repeated entries.
- Improved model performance: Including duplicate samples requires additional calculations and processing power; furthermore, it is computationally wasteful to process samples that have the same values ​​and convey the same information.
- Higher-quality results: By ensuring the data consists only of unique entries—free from repetitions—the information becomes more relevant to the model.
***
The Pandas library provides the `drop_duplicates()` method for removing duplicate samples.
***

```python
no_empty_data.drop_duplicates(inplace=True)
```

```python
no_empty_data.duplicated().sum()
```

# Identification of null data

```python
no_empty_data.isna()
```

```python
no_empty_data.isna().sum()
```

```python
no_empty_data.isna().sum().sum()
```

After executing the cell, we have a total of 114 null values ​​in the dataset. However, this does not represent the number of samples containing null values, as some rows may have more than one null value in different columns.

Therefore, we will perform a filtering operation that returns a *dataframe* containing all rows that have at least one null value in any of the columns.

```python
no_empty_data[no_empty_data.isna().any(axis=1)]
```

Given this information, we need to decide whether to remove samples that contain null values ​​or to fill those nulls with other data.
***
First, let’s consider the option of filling the null values ​​with other data—after all, removing samples also means discarding information that could potentially be relevant.

### Replacement of null data

One of the columns containing null values ​​is `"cliente.tempo_servico"` (*client.time_service*). Since we have `"conta.cobranca.Total"` (*account.charge.Total*) column, and we also have the `"conta.cobranca.mensal"` (*account.charge.monthly*) column, we can use the latter to calculate the duration (months) for which the client has been using the service.

```python
filter = no_empty_data['cliente.tempo_servico'].isna()
```

```python
no_empty_data[filter][['cliente.tempo_servico', 'conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
import numpy as np

np.ceil(5957.90 / 90.45)
```

```python
no_empty_data['cliente.tempo_servico'].fillna(
    np.ceil(
        no_empty_data['conta.cobranca.Total'] / no_empty_data['conta.cobranca.mensal']
    ), inplace=True
)
```

```python
no_empty_data[filter][['cliente.tempo_servico', 'conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
no_empty_data.isna().sum()
```

We now have the `conta.contrato` column, which contains null values ​​that need to be analyzed. Since we have not yet decided to drop any samples, we might consider filling these null values ​​with the most frequent value.

To do this, we need to examine the data within the column. Therefore, we use a Pandas method called `value_counts()`.

```python
no_empty_data['conta.contrato'].value_counts()
```

Upon conducting the analysis, we observed that the most frequent value is `month-to-month`; however, the difference in frequency compared to other values ​​is not particularly significant.

Therefore, we will not impute the missing values ​​using the most frequent value, as doing so would compromise the dataset for the following reasons:

- Data bias: The machine learning model would attempt to identify patterns that are, in fact, incorrect—since they do not reflect the actual data. This can lead to misleading, inaccurate, and unsatisfactory predictions.

- Distortion of results: Introducing an incorrect value causes the model to learn from flawed data, which, in turn, can result in inaccurate predictions.

***


Given that the columns `conta.cobranca_mensal` (*account.charge_monthly*) and `conta.cobranca.Total` (*account.charge.Total*) already have an established relationship—both with each other and with `cliente.tempo_servico` (*client.time_service*)—it may be possible to correctly infer data from them.

What remains is to make a decision regarding the columns `conta.contrato` (*account.contract*), `conta.faturamento_eletronico` (*account.electronic_billing*), and `conta.metodo_pagamento` (*account.payment_method*).

```python
no_empty_data[['conta.contrato', 'conta.faturamente_eletronico' ,'conta.metodo_pagamento']]
```

### Removal of null data

After analyzing the columns containing null values, we determined that it is not possible to accurately infer these missing values ​​based on other data in our *dataframe*. Furthermore, their presence could compromise the efficiency of our machine learning model.

Therefore, we can choose to `drop / exclude` these samples.

```python
columns_drop = ['conta.contrato', 'conta.faturamente_eletronico', 'conta.metodo_pagamento']
```

```python
no_empty_data[columns_drop].isna().any(axis=1)
```

```python
no_empty_data[columns_drop].isna().any(axis=1).sum()
```

We discovered that there are 37 rows containing at least one null value across the three columns. So, we will use another Pandas method called `dropna()`.

```python
no_empty_data.dropna(subset=columns_drop)
```

```python
df_without_null = no_empty_data.dropna(subset=columns_drop).copy()
df_without_null.head()
```

```python
df_without_null.reset_index(drop=True, inplace=True)
df_without_null.head()
```

```python
df_without_null.isna().sum()
```

The output for our new *dataframe* shows that all columns have zero null values. This means there are no longer any null values ​​in our dataset.



***
**But what happened to the null values ​​that were in the 'conta.cobranca.mensal' and 'conta.cobranca.Total' columns?**
***

The values ​​that were null in these two columns were also null in the three columns we were working with. When we removed the samples containing null values ​​in those three columns, we also removed the samples that had null values ​​in `conta.cobranca.mensal` and `conta.cobranca.Total`. In other words, they were the same samples.
