# Understanding the problem

### Churn

Occurs when customers stop doing business with a company.Represents a loss of revenue.I can serve as an indicator of customer satisfaction regarding products/services.


The term "churn" is widely used in the business world and refers to a situation where individuals stop doing business with a company. As data scientists, we know that churn is directly linked to a company's revenue; the more customers who stop doing business, the more money the company loses.


Furthermore, churn can be used as an indicator of customer satisfaction, since a high number of people ceasing to do business may suggest that the service provided is failing to meet customer expectations.

## Transforming data into a tablea

We have learned how to apply the `json_normalize()` method to our dataframes to obtain columns with values ​​in a normalized JSON format. This allows us to expand the keys containing these values ​​into multiple separate columns. However, this approach presents a challenge: having to perform the process manually for each column.

We will now learn a more "Pythonic" way to apply this in a more automated fashion.

### `json_normalize()` method

***
- In a DataFrame ⇾ normalizes only one column

- In a JSON object ⇾ normalizes all nested columns at once
***

```python
import json
```

```python
with open('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/resources/dataset-telecon.json') as f:
  raw_json = json.load(f)
```

```python
raw_json
```

```python
normalized_data = pd.json_normalize(raw_json)
normalized_data.head()
```

```python
normalized_data.to_json('telecon_normalized_data_v1.json', index=False)
```
