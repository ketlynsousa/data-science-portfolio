# Binary categorical variables

Binary categorical variables are variables that have only two mutually exclusive categories or possible responses.

They are used to classify data into opposing groups, such as "yes" or "no," "true" or "false," and "success" or "failure."

## Working with binary variables

In our case, we have the following columns:
- `Churn` (yes/no),
- `cliente.idoso` (0/1),
- `cliente.genero` (male/female),
- `cliente.parceiro` (yes/no),
- `cliente.dependentes` (yes/no).

Among various others, it would be useful to check the possible values ​​for each of our columns.

Our goal is to replace binary values ​​with 0s and 1s. Machine learning models perform much better when provided with numerical data.

```python
mapping = {
    'nao': 0,
    'sim': 1,
    'masculino': 0,
    'feminino': 1
}
```

# Non-binary categorical variables

Non-binary categorical variables are qualitative data that have three or more possible categories, with no inherent order or hierarchy among them.

A classic example is the variable "eye color," which can take on multiple distinct values ​​such as blue, green, brown, or black.

## One Hot Encoder (or Dummy)
Exemple:

```python
s = pd.Series(list('abca'))
s
```

```python
pd.get_dummies(s)
```

The `get_dummies()` method is, in fact, an implementation of the One-Hot Encoding technique. It returns a dataframe containing four samples—indexed from 0 to 3—and three columns: 'a', 'b', and 'c'.

This technique uses categorical variables to create new columns, each named after one of the distinct categories. Since there were three categories, three columns were created in the table.

At index 0, the value in the first column ('a') serves as an answer to the following question: "Is the entry at index 0 equal to 'a'?" Because the value is `True`, the answer is "yes"—which makes sense, as 'a' was the value present at index 0 in the original data.

Since the other two columns do not correspond to the value at index 0, they are populated with the value `False`, signifying "no."

The value at index 1 belongs to category 'b', so a `True` value—corresponding to "yes"—is marked in column 'b'. As before, the remaining columns are set to `False`, representing "no".

This process is repeated for the other samples at each index. This is what the `get_dummies()` method does: it creates new columns and assigns `True` wherever the corresponding index value is indicated in the original dataframe.

### Working with non-binary variables

```python
df_without_id.info()
```

Variables with a dtype set to 'object' will be interpreted as categorical columns.

Shall we apply this to our dataset?

Keep in mind that, in computing, `True` corresponds to the value 1, and `False` to 0.

Therefore, we will use the parameter `dtype=int` to ensure consistency resulting in 0s and 1s within our dataframe.

```python
pd.get_dummies(df_without_id, dtype=int)
```

```python
df_without_id.to_json('subset_without_id.json')
```

```python
df_dummies = pd.get_dummies(df_without_id, dtype=int).copy()
df_dummies.head()
```

```python
df_dummies.columns
```

```python
df_dummies.info()
```

Now, our machine learning model can work with these values ​​in a more optimized and accurate way. If you feed this dataset into a neural network, it will function exclusively with numerical variables—exactly as we intended.

In this way, the model will be able to capture non-linear relationships between the numerical variables and the categorical variables (for which dummy columns have been created).

Our dataset is now ready to be used in a machine learning model!
