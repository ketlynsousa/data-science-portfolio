```python
import pandas as pd
```

```python
data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/exercises/resources/dados.json')
data
```

```python
import json
with open('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/exercises/resources/dados.json', 'r') as f:
  raw_json = json.load(f)
raw_json
```

```python
data = pd.json_normalize(raw_json, record_path='pessoas')
data
```

```python
data = data.explode("telefones")
data.reset_index(inplace=True, drop=True)
data
```

```python
data.info()
```

```python
data['idade'] = data['idade'].astype(int)
```
