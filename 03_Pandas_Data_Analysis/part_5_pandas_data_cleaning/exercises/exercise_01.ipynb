```python
import pandas as pd
import json
```

```python
data = pd.read_json('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/exercises/resources/informacoes.json')
data
```

```python
with open('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/exercises/resources/informacoes.json', 'r') as f:
  raw_json = json.load(f)
```

```python
raw_json
```

```python
normalized_data = pd.json_normalize(raw_json, record_path='enderecos', meta=['nome', 'idade'])
```
