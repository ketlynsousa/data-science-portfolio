# Challenge: processing a database

To facilitate the analysis of course registration data from a platform, you have been provided with a file named [`cursos_cadastrados.json`](https://cdn3.gnarususercontent.com.br/2929-pandas/cursos_cadastrados.json). You have been informed that this dataset may contain issues that could hinder accurate analysis and interpretation.

To address this, you are tasked with identifying and removing entries that exhibit the following problems:

- Null values: missing values ​​in certain columns that must be filled in to ensure an accurate analysis.
- Duplicates: identical records that could skew the analysis, as they represent the same information.
- Empty strings: text values ​​that contain no written content, which can complicate data analysis—especially if a column contains a large number of such entries.
- Data type conversion: some columns in the file may be in the wrong data format (e.g., text instead of numbers), which can interfere with analysis. In such cases, you must convert each column to its correct data type.


Given this context, you will need a systematic approach to clean the data. So, how can this be accomplished?


```python
import pandas as pd
import numpy as np
import json
```

```python
with open('/content/drive/MyDrive/Collab Notebooks - Alura/part_5_pandas_data_cleaning/exercises/resources/cursos_cadastrados.json', 'r') as f:
  data = json.load(f)
df = pd.json_normalize(data)
df
```

1. Identifing and removing samples with null values

```python
df.isnull()
```

```python
df.isnull().sum().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.dropna(inplace=True)
```

2. Identifing and removing samples with duplicate values

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

3. Removing samples with empty strings

```python

df[(df == '').any(axis=1)]
```

```python
df.replace('', pd.NA, inplace=True)
df[df.isnull().any(axis=1)]
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum().sum()
```

```python
df[(df == '').any(axis=1)].sum().sum()
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df['concluintes'] = df['concluintes'].astype(np.int64)
```

```python
df['data_inicio'] = pd.to_datetime(df['data_inicio'])
```

```python
df['data_conclusao'] = pd.to_datetime(df['data_conclusao'])
```

```python
df['preco'] = df['preco'].astype(np.float64)
```
