# Challenge: emissions by economic sector

One of the tasks assigned by the environmental agency's supervisory team was to determine the average annual emissions for each gas. We were able to complete this task by using the `groupby()` and `pivot_table()` methods to generate a table showing emissions over the years for each gas.

The supervisory team now wants to know the average emissions for each economic sector—listed in the "Nível 1 - Setor" (Level 1 - Sector) column—over the years. As a challenge, create a pivot table and a graphical visualization containing this information, using only the `pivot_table()` method to construct the table (i.e., without using the `groupby()` method).

```python
import pandas as pd
```

```python
emissions_per_year = pd.read_csv('/content/emissions_per_year.csv')
```

```python
sectors_emission = emissions_per_year.pivot_table(index='Year', columns='Nível 1 - Setor', values='Emission', aggfunc= 'mean')
sectors_emission
```

```python
sectors_emission.plot(subplots=True, figsize=(10,10));
```
