```python
import pandas as pd
```

```python
gas_per_year = pd.read_csv('/content/gas_per_year.csv')
gas_per_year.head()
```

1. Group the data by the columns "Estado" and "Nível 1 - Setor," calculate the sum of emissions, and store the result in a table.

```python
state_emission_sector = gas_per_year.groupby(['Estado', 'Nível 1 - Setor'])[['Emission']].sum()
state_emission_sector
```

2) Using the table created in activity 1, select the data corresponding to "Energia" under the "Nível 1 - Setor" category.

```python
state_emission_sector.xs('Energia', level=1)
```

3) Using the table created in activity 1, identify the economic activity with the highest emission value for the state of Minas Gerais.

```python
state_emission_sector.xs('MG', level=0)[['Emission']].idxmax()
```

4) Create a table containing the economic activity with the highest emissions for each state.

```python
highest_activity_per_state = state_emission_sector.groupby(level=0).idxmax()
highest_activity_per_state
```

5) Create a table containing the state with the highest emissions for each economic activity.

```python
highest_emission_per_state = state_emission_sector.groupby(level=1).idxmax()
highest_emission_per_state
```
