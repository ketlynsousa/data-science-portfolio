2. Group the data based on the "Nível 1 - Setor" column and locate the data for the "Agriculture and Livestock" (Agropecuária) group.

3. Group the data based on the ""Nível 1 - Setor" column to determine the average emissions for each sector in the year 2021.

4. Group the data based on the ""Nível 1 - Setor" column to determine the total emissions for each sector. Sort the data from the highest to the lowest emission levels.

```python
import pandas as pd
```

```python
gas_emissions = pd.read_csv('/content/emissions_per_year.csv')
gas_emissions.head(4)
```

1. Group the data based on the "Level 1 - Sector" column to view the dictionary containing the keys for the formed groups and the list of indices for each group.

```python
gas_emissions.groupby('Nível 1 - Setor').groups
```

2. Group the data based on the "Nível 1 - Setor" column and locate the data for the "Agriculture and Livestock" (Agropecuária) group.

```python
gas_emissions.groupby('Nível 1 - Setor').get_group('Agropecuária')
```

3. Group the data based on the ""Nível 1 - Setor" column to determine the average emissions for each sector in the year 2021.


```python
gas_emissions[gas_emissions['Year'] == 2021].groupby('Nível 1 - Setor')[['Emission']].mean()
```

4. Group the data based on the ""Nível 1 - Setor" column to determine the total emissions for each sector. Sort the data from the highest to the lowest emission levels.

```python
gas_emissions.groupby('Nível 1 - Setor')[['Emission']].sum().sort_values('Emission', ascending = False)
```
