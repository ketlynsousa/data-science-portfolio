### Parameters 'sep' for the read_csv function

Documentation: https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html

```python
no_comma_data = pd.read_csv(url_2, sep=';')
```

```python
no_comma_data.head()
```

### Parameter 'nrows' to read the first 'x' defined lines.

```python
data_first_rows = pd.read_csv(url, nrows=7)
```

```python
data_first_rows
```

### Parameter to select only columns defined within 'usecols' using their exactly name.

```python
selection_data = pd.read_csv(url, usecols=['Id', 'Year_Birth', 'Income'])
```

```python
selection_data
```

### Parameter to select only columns defined within 'usecols' using their index.

```python
selection_data = pd.read_csv(url, usecols=[0, 1, 4])
```

```python
selection_data
```
