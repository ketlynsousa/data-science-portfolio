## Normalizing JSON files

```python
df_normalized = pd.json_normalize(patients_2_data['Pacientes'])
```

```python
df_normalized
```

### Another way to normalize JSON data

```python
import json

with open('pacientes_2.json', 'r') as f:
  data = json.loads(f.read())

pd.json_normalize(data, record_path='Pacientes', \
                  meta=['Pesquisa', 'Ano'])
```

## Writing JSON files

```python
df_normalized.to_json('normalized_patient_history.json')
```
