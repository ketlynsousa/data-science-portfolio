### Parameters of the read_excel function

Documentation: https://pandas.pydata.org/docs/reference/api/pandas.read_excel.html

```python
emissoes_percapita = pd.read_excel(url, sheet_name='emissoes_percapita')
```

```python
emissoes_percapita
```

```python
data_sources = pd.read_excel(url, sheet_name='fontes')
```

```python
data_sources.head()
```

### Parameter 'usecols' used to specify which columns should be showed.

```python
interval = pd.read_excel(url, sheet_name='emissoes_C02', usecols='A:D')
```

```python
interval
```

### Parameter 'nrows' can be used to define how many rows from the table you want displayed.

```python
interval_2 = pd.read_excel(url, sheet_name='emissoes_C02', usecols='A:D', nrows=10)
```

```python
interval_2
```
