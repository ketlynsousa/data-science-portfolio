## Creating a local databank

```python
import sqlalchemy
```

```python
from sqlalchemy import create_engine, MetaData, Table, inspect
```

```python
engine = create_engine('sqlite:///:memory:')
```

## Writing to a databank

```python
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/refs/heads/main/clientes_banco.csv'
```

```python
import pandas as pd
```

```python
data = pd.read_csv(url)
```

```python
data.head()
```

```python
data.to_sql('clients', engine, index=False)
```

```python
inspector = inspect(engine)
```

```python
print(inspector.get_table_names())
```

## Reading a SQL query

```python
query = 'SELECT * FROM clients WHERE Categoria_de_renda = "Empregado"'
```

```python
employees = pd.read_sql(query, engine)
```

```python
employees
```

```python
employees.to_sql('employees', con=engine, index=False)
```

```python
pd.read_sql_table('employees', engine)
```

```python
pd.read_sql_table('employees', engine, columns=['ID_Cliente',  'Grau_escolaridade', 'Rendimento_anual'])
```

## Updating a databank

```python
query = 'SELECT * FROM clients'
```

```python
pd.read_sql(query, engine)
```

To update the table data, we need to perform the following import:

```python
from sqlalchemy import text
```

After that, to perform the updates, we will implement the following code:

```python
query = 'DELETE FROM clients WHERE ID_Cliente=5008804'
with engine.connect() as conn:
  conn.execute(text(query))
  conn.commit()
```

> The code establishes a connection to a database using SQLAlchemy and executes an SQL query defined in the **query** variable. Within the `with` block—which ensures the connection is automatically closed once the block finishes executing—the query is sent to the database via the `conn.execute(text(query))` method, where `text` is used to ensure the query string is treated as an SQL statement. Following the execution of the query, `conn.commit()` is called to commit the transaction, ensuring that any changes made by the query (such as insertions, updates, or deletions) are permanently applied to the database. <

```python
pd.read_sql_table('clients', engine)
```

```python
query = 'UPDATE clients SET Grau_escolaridade = "Ensino Superior" WHERE ID_Cliente=5008808'
with engine.connect() as conn:
  conn.execute(text(query))
  conn.commit()
```
