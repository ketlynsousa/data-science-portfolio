This DataFrame has 8 columns: id, name, username, email, address, phone, website, and company. Note that the address and company columns contain nested information—that is, the data in these columns is organized into subfields. For example, within the address column, you will find subfields such as street, suite, city, and zipcode. In the company column, the subfields include name (company name), catchPhrase (slogan), and bs (business area/field of activity). This means that, in order to access this information, you must perform a normalization process.

**The challenge** now is to normalize this DataFrame by expanding the `address` and `company` columns into their respective sub-columns, making the data easier to visualize and analyze.

```python
import pandas as pd
import json
import requests
```

```python
user_data = requests.get('https://jsonplaceholder.typicode.com/users')

data_result = json.loads(user_data.text)
```

```python
normalized_user_data = pd.json_normalize(data_result, sep='_')
```
