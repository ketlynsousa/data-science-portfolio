# Challenge: manipulating a database

It is time to test the knowledge you gained during the lesson.

You are tasked with creating a local database of customers for a financial institution. You have a [CSV file](https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv) containing the customer data.

Your mission is:

1. Create a local database using the SQLAlchemy library.
2. Write the data from the CSV file into this local database.
3. Perform three updates to the database:
    - Update the record for the client with ID 6840104, whose annual income has changed to 300,000.
    - Delete the record for the client with ID 5008809, as this individual no longer holds an account with the financial institution.
    - Create a new client record according to the following specifications:
        - Client_ID: 6850985
        - Age: 33
        - Education_level: Doctorate
        - Marital_status: Single
        - Family_size: 1
        - Income_category: Employed
        - Occupation: IT
        - Years_employed: 2
        - Annual_income: 290,000
        Has_car: 0
        - Housing: Owns home/apartment

```python
import pandas as pd
import sqlalchemy
from sqlalchemy import create_engine, inspect, text
```

### Creating local data base

```python
engine = create_engine('sqlite:///:memory:')
```

### Reading the csv file and writing the data into the databas as a table

```python
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv'
```

```python
client_data = pd.read_csv(url)
```

```python
client_data.to_sql('clients', engine, index=False)
```

### Inspecting to ensure table was created successfully

```python
inspector = inspect(engine)
print(inspector.get_table_names())
```

```python
pd.read_sql_table('clients', engine)
```

### Performing updates to the database

Task:
- Update the record for the client with ID 6840104, whose annual income has changed to 300,000.

```python
query = 'SELECT * FROM clients WHERE ID_Cliente = 6840104'
pd.read_sql_query(query, engine)
```

```python
query = 'UPDATE clients SET Rendimento_anual = 300000.00 WHERE ID_Cliente=6840104'

with engine.connect() as conn:
  conn.execute(text(query))
  conn.commit()
```

```python
query = 'SELECT * FROM clients WHERE ID_Cliente = 6840104'
pd.read_sql_query(query, engine)
```
