# Challenge: reading data from another link

It’s time to put the knowledge you gained during the lesson to the test. We have a Google Sheets [link](https://https://docs.google.com/spreadsheets/d/1pvBoLyX8kP0TjtUbadVMGdTl4yzm6bHMThhPiqCVtpw/edit?gid=1214654498#gid=1214654498) containing important data on carbon dioxide emissions worldwide. The dataset, sourced from Kaggle, consists of per capita CO2 emissions for every country in the world from 1990 to 2019.

In this challenge, your task is to read the data from this Google Sheets link and then save the resulting DataFrame in CSV format.

Ready to get started?

```python
import pandas as pd
```

```python
sheet_id = '1pvBoLyX8kP0TjtUbadVMGdTl4yzm6bHMThhPiqCVtpw'

url = f'https://docs.google.com/spreadsheets/d/{sheet_id}/gviz/tq?tqx=out:csv&sheet'
```

```python
data_c02_sheet = pd.read_csv(url)
```

```python
data_c02_sheet.head()
```

```python
data_c02_sheet.to_csv('co2_emission.csv', index=False)
```
