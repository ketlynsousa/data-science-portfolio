## <font color=green> 4. HANDLING EXCEPTIONS

Throughout our journey so far, we have likely encountered various errors and exceptions while executing commands. As a data scientist or programmer, you must remain vigilant regarding these situations to prevent bugs or issues in your code and analyses—problems that could negatively impact the user experience or the efficiency of your work.

Broadly speaking, there are two distinct types of errors: **syntax errors** and **exceptions**.

Exceptions are errors detected during program execution; they disrupt the program's flow and will cause it to terminate if they are not properly handled.

In this section, we will learn how to identify and handle some of these exceptions. However, it is always a good practice to consult the documentation to research and determine which exceptions are relevant to your specific projects.

## 4.1 Handling Exceptions

Exception handling helps by establishing an alternative flow for code execution, preventing processes from being interrupted unexpectedly.

There is a wide range of exceptions; depending on the desired behavior and the specific errors you wish to address, you can define a path for the user to follow or provide further details regarding a given exception.

- Exception Hierarchy (https://docs.python.org/3/library/exceptions.html#exception-hierarchy)

### Try ... Except


#### **Situation 15:**

You have created a function to calculate a student's average grade in a given subject by passing in a list of the student's grades.

You need to handle two scenarios:
- If the list contains a non-numeric value, the average calculation should not be performed, and the message "Could not calculate the student's average. Only numeric values ​​are accepted!" should be displayed.
- If the list contains more than 4 grades, a **ValueError** exception should be raised with the message "The list cannot contain more than 4 grades."

A message stating "The query has ended!" must be displayed, regardless of whether an exception is raised or not.

**Shall we solve this challenge?**

````python
def average(lst: list=[0]) -> float:
  ''' Function to calculate the average of grades provided in a list

  lista: list, default [0]
  List containing the grades for which the average is to be calculated
  return = calculo: float
  Calculated average
'''

  calculation = sum(lst) / len(lst)

  if len(lst) > 4:
    raise ValueError('The list cannot contain more than 4 grades.')

  return calculation
````

````python
grades = [6, 7, 8, 8, 9]
result = average(grades)
result
````

````python
grades = [6, 7, 8, 8, 9, "8"]
result= average(grades)
result
````
