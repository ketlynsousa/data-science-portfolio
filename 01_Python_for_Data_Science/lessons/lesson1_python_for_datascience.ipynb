# Conditional structures

## `IF` and `ELSE`

`if` and `else` are two conditional structures. The `if` statement will execute a block of code if the specified condition is **true**. The `else` statement applies in cases where the `if` condition is **false**.

`if` is a keyword in Python that means "if." It is used to create a conditional structure, which allows you to check whether a given condition is true or false and then execute a specific block of code based on the result of that check. The syntax for using `if` is:



```
if condition:
# do something
```


````python
if 2 > 7:
  print('True condition')
print('Out of block')
````

In Python, `else` is used in conjunction with the `if` keyword to form a conditional structure. The syntax for using `else` is:

```
if condition:
# code to execute if true
else:
# code to execute if false
```

The `else` block is executed when the condition checked by the `if` statement evaluates to `False`.

---
**Situation:**

We will receive a student's average grade, and we need an algorithm that performs an analysis to determine whether the student is **Approved** or **Failed**, displaying a message with the result. To be approved, the average grade must be 6.0 or higher.

````python
average = float(input('Enter the average: '))

if average >= 6.0:
  print('Approved')
else:
  print('Failed')
````

Our educational institution has now issued an official statement indicating that students with an average grade between 4.0 and 6.0 are eligible to take **remedial** courses during the break to improve their grades.

We can therefore use a series of `if` statements to structure this new condition.

````python
average = float(input('Enter the average: '))

if average >= 6.0:
  print('Approved')
if 4.0 <= average < 6.0:
  print('Recovery')
if average < 4.0:
  print('Failed')
````

Note that in cases involving three scenarios—like this one—we need to clearly define our conditions. Since an `else` clause is used at the end, it will treat any remaining outcome as the "false" case, which could potentially lead to two (or more) blocks of code being executed.

For example:

````python
average = float(input('Enter the average: '))

if average >= 6.0:
  print('Approved')
if 4 <= average < 6.0:
  print('Recovery')
else:
  print('Failed')
````

## `ELIF`

`elif` is a keyword in Python that stands for "else, if" and can be thought of as a combination of `else` and `if`. It is used in conjunction with the `if` keyword to form a chained conditional structure.



The syntax for using `elif` is:

```
if condition1:
# do something
elif condition2:
# do something else
elif condition3:
# do yet another thing
else:
# do something different
```

`elif` allows you to check multiple conditions in a chained manner, saving space in your code. If the first condition evaluates to `False`, the Python interpreter will evaluate the next condition in the `elif` block. This continues until a condition evaluates to `True` or the `else` block is reached. If none of the conditions evaluate to `True`, the code within the `else` block will be executed.

Let’s use the same case as before:

````python
average = float(input('Enter the average: '))

if average >= 6.0:
  print('Approved')
elif 4 <= average < 6.0:
  print('Recovery')
else:
  print('Failed')
````

## Operators

When constructing commands, we sometimes need to develop a more elaborate conditional expression, which may require the inclusion of certain logical operators.

### `AND`, `OR`, `NOT`

The logical operators `and`, `or`, and `not` are used to combine logical expressions in Python. They are frequently used in conjunction with `if` to create more complex conditional structures.


- `AND` is used to check whether two conditions are true. The logical expression¹ `x and y` evaluates to `True` only if **both conditions `x` and `y` are true**, and to `False` otherwise.

- `OR` is used to check whether at least one of the conditions is true. The logical expression `x or y` evaluates to `True` **if at least one of the conditions, `x` or `y`, is true**, and to `False` if both are false.

- `NOT` is used to **negate a condition**. The logical expression `not x` evaluates to `True` if the condition `x` is false, and to `False` if the condition `x` is true.

¹ A logical expression is a statement that can be evaluated as either true or false. It consists of logical operands² and logical operators³, which are used to combine multiple logical expressions into a single expression.

² Logical operands are the elements that are compared or evaluated within a logical expression. They are typically true or false values, but they can also be more complex logical expressions. In Python, the logical operands are the values ​​`True` and `False`.

³ Logical operators are the symbols or keywords used to combine multiple logical expressions into a single expression. In Python, the logical operators are `and`, `or`, and `not`, as well as the keywords `if`, `elif`, and `else`.

````python
t1 = t2 = True
f1 = f2 = False
````

````python
if t1 and f2:
  print('true expression')
else:
  print('false expression')
````

````python
if t1 or f2:
  print('true expression')
else:
  print('false expression')
````

````python
if not f1:
  print('true expression')
else:
  print('false expression')
````

# Repetition structures

When we need to execute the same block of commands multiple times, doing so manually is not very efficient.

Let’s imagine a scenario where we need to collect and print the average of two grades for **3 students**:

````python
grade_1 = float(input('Enter the 1st grade: '))
grade_2 = float(input('Enter the 2nd grade: '))
print(f'Average: {(grade_1+grade_2)/2}')

grade_1 = float(input('Enter the 1st grade: '))
grade_2 = float(input('Enter the 2nd grade: '))
print(f'Average: {(grade_1+grade_2)/2}')

grade_1 = float(input('Enter the 1st grade: '))
grade_2 = float(input('Enter the 2nd grade: '))
print(f'Average: {(grade_1+grade_2)/2}')
````

Now, let’s imagine a scenario involving not just 3 students, but 100. It wouldn’t be practical to repeat the same code 100 times; instead, we want to **execute the same code 100 times**.

We can achieve this kind of repetition using loops!

## `WHILE`

The `while` loop is a repetition control structure in Python that allows a block of code to be executed repeatedly as long as a given condition remains true. Its structure is:



```
while condition:
# code block
```

Let’s build an example using a counter that goes from 1 to 10.


````python
counter = 1
while counter <= 10:
  print(counter)
  counter += 1
````

Now, let’s collect the grades and averages for each student within the `while` loop. We will provide an example using 3 averages.

````python
counter = 1
while counter <= 3:
  grade_1 = float(input('Enter the 1st grade: '))
  grade_2= float(input('Enter the 2nd grade: '))

  print(f'Average: {(grade_1+grade_2)/2}')
  counter += 1
````

## `FOR`

The `for` loop is a type of flow control structure in Python that allows you to iterate over a collection of elements. Its structure is as follows:

```
for element in collection:
# code to be executed for each element
```

The `for` loop iterates over each element in the specified collection and executes the block of code within the loop for each one. When the loop reaches the end of the collection, it terminates, and the program continues execution after the loop.



The set can be generated using the [`range()`](https://docs.python.org/3/library/functions.html#func-range) function, which is capable of producing a sequence of integers. The structure of this function is:

```
range(start, end, step)
```

According to the documentation, `range()` generates a sequence of integers starting from the value of the `start` parameter up to the value of the `end` parameter, based on the value of the `step` parameter. If `start` is not specified, the default value is 0. If `step` is not specified, the default value is 1.




Let's create the same `while` counter, but this time using `for`.

````python
for counter in range(1, 11):
  print(counter)
````

````python
for counter in range(1, 4):
  grade_1 = float(input('Enter the 1st grade: '))
  grade_2= float(input('Enter the 2nd grade: '))

  print(f'Average: {(grade_1+grade_2)/2}')
````
