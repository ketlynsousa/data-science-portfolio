```

print(f'{num_2} is an integer?:', num_2 // 1 == num_2)
```
Output:

```

Given that the distances from Recife to each of these cities are approximately `[850, 800, 300, 550]` km, create three functions: the first to calculate hotel expenses (`hotel_expenses`), the second to calculate fuel costs (`fuel_cost`), and the third to calculate expenses for sightseeing and meals (`extra_costs`).

To test your code, simulate a 3-day trip to Salvador, departing from Recife. Account for a round-trip journey by car.
```
"Based on the defined expenses, a [days]-day trip to [city] departing from Recife would cost [total_expenses] BRL."

````python
distances = [850, 800, 300, 550]
tour = [200, 400, 250, 300]
km_per_liter = 14
fuel_price = 5

def hotel_cost(days):
  daily_rate = 150 * days
  return daily_rate

def fuel_cost(city):
  if city == 'Salvador':
    return (2 * distances[0] * fuel_price) / km_per_liter
  elif city == 'Fortaleza':
    return (2 * distances[1] * fuel_price) / km_per_liter
  elif city == 'Natal':
    return (2 * distances[2] * fuel_price) / km_per_liter
  elif city == 'Aracaju':
    return (2 * distances[3] * fuel_price) / km_per_liter

def tour_cost(city, days):
  if city == 'Salvador':
    return tour[0] * days
  elif city == 'Fortaleza':
    return tour[1] * days
  elif city == 'Natal':
    return tour[2] * days
  elif city == 'Aracaju':
    return tour[3] * days

days = int(input("How many days (number of nights)? "))
city = input("Which city? [Salvador, Fortaleza, Natal, or Aracaju]: ")

total_expenses = hotel_cost(days) + fuel_cost(city) + tour_cost(city, days)
print(f"Based on the defined expenses, a {days}-day trip to {city} departing from Recife would cost {round(total_expenses, 2)} BRL.")

````

Write a program that prompts the user to enter two `float` numbers and calculates the result of dividing them. The code must include error handling to indicate the type of error that occurs if the division cannot be performed.

Test the program by setting the second numerical input to 0, and also test it by entering text characters to observe the types of errors that arise.

````python
try:
  num1 = float(input('Enter the first number: '))
  num2 = float(input('Enter the second number: '))
  result = num1 / num2
except ZeroDivisionError as e:
  print(f'Error: {e}. It is not possible to divide a number by ZERO.')
else:
  print(result)
finally:
  print('Canculation complete!')
````

Create a function that takes a list as a parameter and converts all values ​​in the list to `float`. The function must include error handling that indicates the type of error generated, and it should return the list if no errors occur. Finally, it must include a `finally` clause that prints the text: 'End of function execution'.

````python
def converter(data: list=[0]) -> list:
  try:
    float_list = [float(value) for value in data]
  except Exception as e:
    print(f'{type(e)} Error: {e}')
  else:
    return float_list
  finally:
    print('End of function execution')

numbers = [2, 7, 9, 6]
decimal = converter(numbers)
decimal
````

Create a function that takes two lists as parameters and groups their elements one by one, forming a list of 3-element tuples. In each tuple, the first and second elements should be the values ​​at position `i` from the respective lists, and the third element should be the sum of those two values. The function must include error handling that identifies the type of error encountered and returns the list of tuples as the result. If the lists provided as parameters have different lengths, the function should raise an `IndexError` with the message: 'The number of elements in each list is different.'

Data for testing the function:

Values ​​without errors:
```
list1 = [4, 6, 7, 9, 10]
list2 = [-4, 6, 8, 7, 9]
```
Lists with different lengths:
```
list1 = [4, 6, 7, 9, 10, 4]
list2 = [-4, 6, 8, 7, 9]
```
Lists with incompatible values:
```
list1 = [4, 6, 7, 9, 'A']
list2 = [-4, 'E', 8, 7, 9]
```

````python
def calculation_of_lists(list1, list2):
  try:
    if len(list1) == len(list2):
      data = [([list1[i], list2[i], list1[i] + list2[1]]) for i in range(len(list1))]
    else:
      raise IndexError('The number of elements in each list is different.')
  except Exception as e:
    print(type(e), f'Error: {e}')
  else:
    return data

list1 = [4,6,7,9,'A']
list2 = [-4,'E',8,7,9]
calculation_of_lists(list1=list1, list2=list2)
````

#### Question 5

You have been tasked with developing code that calculates the scores of students at an educational institution based on their answers to a test.

This code should be tested using a sample of 3 students, represented by a list of lists, where each inner list contains the answers to 5 objective questions for a given student. Each question is worth one point, and the possible answer choices are A, B, C, or D.

If any answer in one of the tests is not among the valid options, you must raise a **ValueError** with the message: `"The answer [answer] is not a valid option"`. The scores for the 3 students should only be calculated if all answers across all tests are either A, B, C, or D.

If no exception is raised, the program should output a list containing the scores for each test.

The data to be used for testing the code is as follows:

- **Answer key:**
```
result = ['D', 'A', 'B', 'C', 'A']
```
Below are two lists of lists that you can use for testing:

- **Tests without an exception:**
```
tests_without_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']]
```
- **Tests with an exception:**
```
testes_with_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'E', 'A'], ['D', 'B', 'A', 'C', 'A']]
```

> Tip: To check if an entry in the list is not among the valid options, use the structure `list[i] not in ['A', 'B', 'C', 'D']` (For example, `1 not in [2, 3, 4]`... Output: `True`)

````python
# Function for grading tests
result = ['D', 'A', 'B', 'C', 'A']
def corrector(tests: list):
  scores = []
  try:
    for test in tests:
      score = 0
      for i in range(len(test)):
        if test[i] not in ['A', 'B', 'C', 'D']:
          raise ValueError(f'The answer {test[i]} is not a valid option.')
        elif test[i] == result[i]:
          score += 1
    scores.append(score)
  except Exception as e:
    print(e)
  else:
    return scores
````

````python
# Testing with an example that does not raise an exception
tests_without_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']]
corrector(tests_without_ex)
````

````python
# Testing with the example that throws an exception
tests_with_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'E', 'A'], ['D', 'B', 'A', 'C', 'A']]
corrector(tests_with_ex)
````

#### Question 7
You have been hired as a data scientist to assist a laboratory conducting experiments on the behavior of a fungal culture. The laboratory needs to constantly evaluate the ratio (division) between pressure and temperature data from a controlled environment—collected during the experiments—to determine the optimal conditions for testing.

To meet this requirement, you must create a function called `divide_columns` that takes pressure and temperature data (provided as lists) and generates a new list containing the results of the division. The function should accept these two lists as parameters and include error handling for at least two types of exceptions:
- Check whether the lists are of the same length (`ValueError`).
- Check for any instances of division by zero (`ZeroDivisionError`).

To test the function, you will perform the division using two lists of data collected during the experiment, representing the pressure and temperature values ​​in the controlled environment.

Use the following data for your tests:

- **Data without exceptions:**
```
pressures = [100, 120, 140, 160, 180]
temperatures = [20, 25, 30, 35, 40]
```
- **Data with exceptions:**
1. `ZeroDivisionError` case:
```
pressures = [60, 120, 140, 160, 180]
temperatures = [0, 25, 30, 35, 40]
```
2. `ValueError` case:
```
pressures = [100, 120, 140, 160]
temperatures = [20, 25, 30, 35, 40]
```

> Tip: You can use `zip()` to pair the elements from the first list with those from the second list. Create a try-except structure so that, if an exception is raised, we can see the type of error in the output.

````python
# Testing with an example that does not throw an exception
def divide_columns(pressure: list, temperature: list):
  try:
    if len(pressure) != len(temperature):
      raise IndexError('The number of elements in each list is different.')
    result = [round(a / b, 2) for a, b in zip(pressure, temperature)]
  except IndexError as e:
    print(e)
  except ZeroDivisionError as e:
    print(f'{e}: The 2nd list cannot contain a value equal to 0')
  else:
    return result
````

````python
# Testing with an example that does not raise an exception
pressures = [100, 120, 140, 160, 180]
temperatures = [20, 25, 30, 35, 40]

divide_columns(pressures, temperatures)
````

````python
# Testing with an example that raises an exception (ZeroDivisionError)
pressures = [60, 120, 140, 160, 180]
temperatures = [0, 25, 30, 35, 40]

divide_columns(pressures, temperatures)
````

````python
# Testing with an example that raises an exception (ValueError)
pressures = [100, 120, 140, 160]
temperatures = [20, 25, 30, 35, 40]

divide_columns(pressures, temperatures)
````
