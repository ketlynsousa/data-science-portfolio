#### Question 5
Write a program that calculates the factorial of an integer provided by the user. The factorial of an integer is the product of that number and all the integers preceding it, down to 1. For example, the factorial of 5 is 5 x 4 x 3 x 2 x 1 = 120.

````python
# Pedir o número
num = int(input('Informe um número inteiro: '))

# Inicializar o cálculo
fatorial = 1

# nosso contador inicializa com o número máximo
# e será feita uma contagem decrescente com o operador -=
i = num
while i > 0:
    # queremos multiplicar agora o valor do fatorial pelo num
    # e todos os números abaixo dele até 1
    fatorial *= i
    i -= 1

# Imprimir o cálculo do fatorial
print(f'Fatorial de {num} é {fatorial}')
````

````python
# Ask for a number
num = int(input('Enter an integer: '))

# Initialize the calculation
factorial = 1

# Our counter starts at the maximum number
# and will count down using the -= operator
i = num
while i > 0:
# We multiply the current factorial value by the number
# and all subsequent numbers down to 1
  factorial *= i
  i -= 1

# Print the factorial result
print(f'The factorial of {num} is {factorial}')
````

#### Question 8
Let’s analyze the age distribution of pensioners at a pension fund company. Write a program that reads the ages of an unspecified number of clients and displays their distribution across the following intervals: [0–25], [26–50], [51–75], and [76–100]. Terminate data entry by entering a negative number.

````python
age = int(input('Enter the age (or a negative number to finish): '))

counter_0_25 = counter_26_50 = counter_51_75 = counter_76_100 = 0

while age >= 0:
  if age >= 0 and age <= 25:
    counter_0_25 += 1
  elif age >= 26 and age <= 50:
    counter_26_50 += 1
  elif age >= 51 and age <= 75:
    counter_51_75 += 1
  elif age >= 76 and age <= 100:
    counter_76_100 += 1

  age = int(input('Enter the age (or a negative number to finish): '))

print('Age distribution:')
print('[0-25]:', counter_0_25)
print('[26-50]:', counter_26_50)
print('[51-75]:', counter_51_75)
print('[76-100]:', counter_76_100)
````

#### Question 9
In an election for a management position at a company with 20 employees, there are four candidates. Write a program to determine the winner of the election. The voting process took place as follows:
- Each employee voted for one of the four candidates (represented by the numbers 1, 2, 3, and 4).
- Null votes (represented by the number 5) and blank votes (represented by the number 6) were also recorded.

At the end of the voting, the program must display the total number of votes for each candidate, as well as the number of null votes and blank votes. Furthermore, it must calculate and display the percentage of null votes and the percentage of blank votes relative to the total number of votes cast.

````python
candidate_1 = candidate_2 = candidate_3 = candidate_4 = 0
null_votes = blank_votes = 0

for x in range(0, 20):
  vote = int(input('Enter your vote: '))
  if vote == 1:
    candidate_1 += 1
  elif vote == 2:
    candidate_2 += 1
  elif vote == 3:
    candidate_3 += 1
  elif vote == 4:
    candidate_4 += 1
  elif vote == 5:
    null_votes += 1
  elif vote == 6:
    blank_votes += 1
  else:
    print('Invalid vote.')

print(f'Votes for candidate 1: {candidate_1}')
print(f'Votes for candidate 2: {candidate_2}')
print(f'Votes for candidate 3: {candidate_3}')
print(f'Votes for candidate 4: {candidate_4}')
print(f'Null votes: {null_votes}')
print(f'Blank votes: {blank_votes}')
print(f'Percentage of null votes: {null_votes * 20 / 100}%')
print(f'Percentage of blank votes: {blank_votes * 20 / 100}%')
````

#### Question 5
Write a program that, upon receiving an input number, creates a list containing all prime numbers between 1 and the number entered.

````python
number = int(input('Enter an integer: '))
list_of_prime = []

for num in range(2, number):
  prime_number = True

  for divisible in range(2, num):
    if num % divisible == 0:
      prime_number = False
      break
  if prime_number:
    list_of_prime.append(num)

print(f'List of prime numbers: {list_of_prime}')
````

#### Question 7
For a study on the rate of bacterial multiplication in a colony, the number of bacteria present each day—based on a daily doubling process—was recorded as follows: `[1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]`. Given these values, write code that generates a list containing the daily percentage growth of the bacteria, by comparing the number of bacteria on each day to the number from the previous day. Tip: to calculate the percentage growth, use the following equation: `100 * (current_sample - previous_sample) / (previous_sample)`.

````python
bacteria_colony = [1.2, 2.1, 3.3, 5.0, 7.8, 11.3, 16.6, 25.1, 37.8, 56.9]
growth_porcentage = []

for i in range(1, len(bacteria_colony)):
  porcentage = 100 * (bacteria_colony[i] - bacteria_colony[i-1]) / bacteria_colony[i-1]
  growth_porcentage.append(porcentage)

print(f'Growth percentage: \n{growth_porcentage}')
````

#### Question 13
Employees in a department at your company are set to receive a bonus equal to 10% of their salary, in recognition of the team's excellent performance. The finance department has asked for your help in assessing the financial impact of these bonuses. You have been provided with a list of the salaries that will qualify for the bonus: `[1172, 1644, 2617, 5130, 5532, 6341, 6650, 7238, 7685, 7782, 7903]`. Note that no employee's bonus may be less than 200. Write code to create a dictionary where each salary serves as a key and its corresponding bonus as the value. Then, report the total expenditure on these bonuses, the number of employees who received the minimum bonus amount, and the highest bonus value awarded.

````python
salaries = [1172, 1644, 2617, 5130, 5532, 6341, 6650, 7238, 7685, 7782, 7903]

dic_bonuses = {}
total_bonus = minimum_bonus = higher_bonus = 0

for salary in salaries:
  bonus = salary * 0.1
  if bonus < 200:
    bonus = 200
  dic_bonuses[salary] = bonus

for bonus in dic_bonuses.values():
  if bonus == 200:
    minimum_bonus += 1

  if bonus > higher_bonus:
    higher_bonus = bonus

  total_bonus += bonus

print(f'Bonuses: {dic_bonuses}')
print(f'Total expenses on bonuses: {total_bonus}')
print(f'Number of employees who received the minimum bonus: {minimum_bonus}')
print(f'Highest bonus amount: {higher_bonus}')
````

#### Question 14
A team of data scientists is studying biological diversity in a forest. The team collected information on the number of plant and animal species in each area of ​​the forest and stored this data in a dictionary. In this dictionary, the key represents the forest area, and the associated list contains the number of plant and animal species, respectively.
```
{'North Area': [2819, 7236],
'East Area': [1440, 9492],
'South Area': [5969, 7496],
'West Area': [14446, 49688],
'Central Area': [22558, 45148]}
```
Write code to calculate the average number of species per area and identify the area with the highest biological diversity. Tip: use the built-in functions [`sum()`](https://docs.python.org/3/library/functions.html#sum) and [`len()`](https://docs.python.org/3/library/functions.html#len).

````python
data = {
'North Area': [2819, 7236],
'East Area': [1440, 9492],
'South Area': [5969, 7496],
'West Area': [14446, 49688],
'Central Area': [22558, 45148]
}

highest_diversity = ''
sum_of_averages = highest_sum = 0

for area, species in data.items():
  sum_species = sum(species)
  average = sum_species / len(species)

  print(f'The {area} has an average of {average} species')

  if sum_species > highest_sum:
    highest_sum = sum_species
    highest_diversity = area

  sum_of_averages += average

print('-' * 50)
overall_average = sum_of_averages / len(data)
print(f'Overall average of species: {overall_average}')
print(f'Area with the highest biological diversity: {highest_diversity}')
````

#### Question 15
Your company's HR department has asked for your help in analyzing the ages of employees across 4 different departments. They have provided the following data:
```
{'Sector A': [22, 26, 30, 30, 35, 38, 40, 56, 57, 65],
'Sector B': [22, 24, 26, 33, 41, 49, 50, 54, 60, 64],
'Sector C': [23, 26, 26, 29, 34, 35, 36, 41, 52, 56],
'Sector D': [19, 20, 25, 27, 34, 39, 42, 44, 50, 65]}
```
Given that each department has 10 employees, write code that calculates the average age for each department, the overall average age across all departments, and the number of individuals whose age is above this overall average.

````python
data = {
'Sector A': [22, 26, 30, 30, 35, 38, 40, 56, 57, 65],
'Sector B': [22, 24, 26, 33, 41, 49, 50, 54, 60, 64],
'Sector C': [23, 26, 26, 29, 34, 35, 36, 41, 52, 56],
'Sector D': [19, 20, 25, 27, 34, 39, 42, 44, 50, 65]
}

sum_of_averages = above_average_age_count = 0

for sector, employees in data.items():
  sum_of_ages = sum(employees)
  average = sum_of_ages / len(employees)

  print(f'{sector} has an average age of {average} years.')

  sum_of_averages += average
  for age in employees:
    if age > average:
      above_average_age_count += 1

print('-' * 50)
overall_average = sum_of_averages / len(data)
print(f'The overall average age across all sectors is {overall_average}')
print(f'The number of people with an age above the average is: {above_average_age_count}')
````
